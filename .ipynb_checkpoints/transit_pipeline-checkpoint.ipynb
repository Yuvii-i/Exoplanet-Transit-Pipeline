```python
import lightkurve as lk
import numpy as np
import matplotlib.pyplot as plt
import batman
```

```python
%matplotlib inline
```

```python
search_result = lk.search_lightcurve("WASP-17", mission="TESS")
lc_collection = search_result[[0, 1, 3]].download_all()
lc = lc_collection.stitch()
```

```python
lc_clean = lc.remove_outliers(sigma=5).flatten(window_length=1440)
```

```python
pg = lc_clean.to_periodogram(method='bls', minimum_period=1, maximum_period=27, frequency_factor=150)
```

```python
pg.period_at_max_power
```

```python
lc_folded = lc_clean.fold(period=pg.period_at_max_power, epoch_time=pg.transit_time[0]).bin(time_bin_size=0.0104)
```

```python
params = batman.TransitParams()
params.per = 3.735
params.t0 = 1.42
params.rp = 0.111
params.inc = 86.83
params.a = 7.03
params.w = 90
params.ecc = 0
params.u = [0.29, 0.31]
params.limb_dark = 'quadratic'

t = lc_folded.time.value
m = batman.TransitModel(params, t)

flux = m.light_curve(params)
```

```python
ax = lc_folded.plot()
ax.set_ylim(0.98, 1.002)

ax.plot(t, flux)
plt.show()
```
